```python
import pandas as pd

df = pd.read_csv("sales_data/train.csv")

df.head()
```

```python
print("Rows and columns:", df.shape)
print("Total records:", len(df))
print("Total columns:", len(df.columns))
```

```python
# Check missing values in each column
missing_values = df.isnull().sum()

print(missing_values)
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
df["Postal Code"] = df["Postal Code"].fillna(0).astype(int)

print("Missing values after handling:")
print(df.isnull().sum().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Count duplicate rows
print("Duplicates before cleaning:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Check again
print("Duplicates after cleaning:", df.duplicated().sum())
```

```python
# Convert date columns to datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

# Check the data types
print(df[["Order Date", "Ship Date"]].dtypes)
```

```python

# Convert date columns to datetime format
df["Order Date"] = pd.to_datetime(
    df["Order Date"], dayfirst=True, errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"], dayfirst=True, errors="coerce"
)

# Check the data types
print(df[["Order Date", "Ship Date"]].dtypes)

# Display the first five rows
df[["Order Date", "Ship Date"]].head()
```

```python

# Step 8: Convert date columns to datetime format

df["Order Date"] = pd.to_datetime(
    df["Order Date"], format="mixed", dayfirst=True, errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"], format="mixed", dayfirst=True, errors="coerce"
)

# Check the data types
print(df[["Order Date", "Ship Date"]].dtypes)
```

```python

# Check missing values
print("Total missing values:", df.isnull().sum().sum())

# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Check the dataset size
print("Rows and columns:", df.shape)
```

```python

# Save the cleaned dataset
df.to_csv("sales_data_cleaned.csv", index=False)

print("Cleaned dataset exported successfully!")
```
